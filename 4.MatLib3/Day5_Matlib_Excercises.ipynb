```python
# 9.Create a bar chart with different colors for each bar. Add labels and titles to a bar chart using xlabel(), ylabel() 
# and title().Display values above each bar using plt.text().
Datax=['red','blue','green','violet','purple']
Datay=[10,20,30,40,50]
plt.xlabel('Colour Names')
plt.ylabel('Values')
plt.title('Colour Demonstration')
plt.bar(Datax,Datay,color=['red','blue','green','violet','purple'])
for x, y in zip(Datax, Datay):
    plt.text(x, y, str(y))

plt.show()
```

```python
# 10.Create a grouped bar chart comparing the marks of students in Maths, Science and English.
students_1 = ["Student A", "Student B", "Student C", "Student D", "Student E"]
maths = [85, 78, 92, 74, 88]
science = [80, 82, 89, 76, 91]
english = [88, 75, 85, 90, 79]
barwidth=0.25
bar1=np.arange(len(maths))
bar2=[x + barwidth for x in bar1]
bar3=[x + barwidth for x in bar2]
plt.bar(bar1,maths,width=barwidth,label='Maths')
plt.bar(bar2,science,width=barwidth,label='Science')
plt.bar(bar3,english,width=barwidth,label='English')
plt.xticks(bar1,students_1)
plt.xlabel("Students")
plt.ylabel("Marks scored in Maths,Science & English")
plt.legend(bbox_to_anchor=(1, 0.2))
```
