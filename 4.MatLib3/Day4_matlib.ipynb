```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
company=['GOOGL','AMZN','MSFT','FB']
revenue=[90,136,89,27]
profit=[40,2,34,12]
xpos=np.arange(len(company))
plt.bar(xpos-0.2,revenue,width=0.4,label='Revenue')
plt.bar(xpos+0.2,profit,width=0.4,label='Profit')
plt.xlabel('Company')
plt.ylabel('Revenue')
plt.xticks(xpos,company)
plt.title('Company Revenues')
plt.legend()
plt.show()
```

```python
company=['GOOGL','AMZN','MSFT','FB']
revenue=[90,136,89,27]
profit=[40,2,34,12]
xpos=np.arange(len(company))
plt.barh(xpos-0.2,revenue,height=0.4,label='Revenue')
plt.barh(xpos+0.2,profit,height=0.4,label='Profit')
plt.xlabel('Company')
plt.ylabel('Revenue')
plt.yticks(xpos,company)
plt.title('Company Revenues')
plt.legend()
plt.show()
```

```python
# Creating dataset
cars = ['AUDI', 'BMW', 'FORD',
        'TESLA', 'JAGUAR', 'MERCEDES']

data = [23, 17, 35, 29, 12, 41]

# Creating plot
fig = plt.figure(figsize=(15,7))
plt.pie(data, labels=cars)


# show plot
plt.show()
```

```python
expense_values=[1400,600,300,410,250]
expense_labels=['Home rent','Food', 'Phone or internet bill','Car','Other utilities']
plt.pie(expense_values,labels=expense_labels,
        shadow=True,autopct='%0.1f%%',radius=2,
        explode=[0.2,0,0.5,0,0],counterclock=False) #auto pct - floating points
plt.axis('equal')
plt.savefig("Test_fig2.pdf",bbox_inches='tight',pad_inches=3)
plt.show()

```
