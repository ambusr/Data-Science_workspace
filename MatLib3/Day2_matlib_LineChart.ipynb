Line Chart Advanced


```python
x = np.array([1, 2, 3, 4])
y = x * 2

x1 = [2, 4, 6, 8]
y1 = [3, 5, 7, 9]

# 👇 Plot both lines without calling plt.figure()
plt.plot(x, y, label='Line 1')
# plt.figure()
plt.plot(x1, y1, '-.', label='Line 2')
plt.plot(x, y1, '-.', label='Line 3')
plt.plot(x1, y, '-.', label='Line 4')

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Both Lines in One Chart")
plt.fill_between(x1, y, y1, color='red', alpha=1)
plt.legend(loc='best',shadow=True,fontsize='large') #best, upper right, lower right etc
plt.grid()
# plt.show()
```

```python
days=[1,2,3,4,5,6,7]
max_t=[50,51,52,48,47,49,46]
min_t=[43,42,40,44,33,35,37]
avg_t=[45,48,48,46,40,42,41]

plt.xlabel('Days')
plt.ylabel('Temperature')
plt.title('Weather')
plt.plot(days,max_t,label='Maximum Temp')
plt.plot(days,min_t,label='Minimum Temp')
plt.plot(days,avg_t,label='Average temp')
plt.legend(loc='upper right',fontsize='small')

plt.show()
```
