Data Cleaning practice Using Pandas.
Customer Dataset Cleaning
Create a dataset containing:

Customer_ID,Customer_Name,,Age,Gender,Email,Phone,City,Purchase_Amount,Purchase_Date

Perform the following:
1.Handle missing values.
2.Remove duplicate customers.
3.Correct invalid ages.
4.Clean customer names.
5.Validate email addresses.
6.Clean phone numbers.
7.Standardize city names.
8.Convert purchase dates into datetime.
9.Detect purchase amount outliers.
10.Export the final cleaned dataset.

```python
import pandas as pd
import numpy as np
df=pd.read_csv("customer_data_cleaning_practice.csv")
```

```python
df.info()
```

```python
display(df[df.duplicated()])

```

Droping duplicates & checking shape

```python
df[df.duplicated(subset=['Customer_ID'])==True] # Checking for any duplicate Cust ID
```

```python
df.drop_duplicates(keep='first',inplace=True)
df.shape
```

Droping rows with NAN values in Customer ID & Customer name

```python
df.dropna(subset=['Customer_ID','Customer_Name'],inplace=True)
df.shape
```

Clearing invalid Ages

```python
# df.loc[(df['Age'] < 1) | (df['Age'] > 100)]
print(df['Age'].dtype)
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce') #converting ages to numeric, errror - coerce make invalid values to NAN
print(df['Age'].dtype)
```

```python
df.loc[(df['Age'] < 1) | (df['Age'] > 100)] #check for invalid ages
```

```python
df.loc[(df['Age']<1) | (df['Age']>100),'Age']= np.nan # Assigning invalid values to nan
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median()) #Assigning median values to missing age
display(df.loc[(df['Age'] < 1) | (df['Age'] > 100)])
display(df[df['Age'].isna()==True])
```

Fill missing Gender with the most frequent gender (mode)

```python
print(df.Gender.dtype) #check data type
df.Gender=df.Gender.str.title()
df.loc[~((df.Gender=='Male')|(df.Gender=='Female'))]
```

```python
df.loc[~((df.Gender=='Male')|(df.Gender=='Female')),'Gender']=np.nan #Assign NAN to invlaid gender
df.Gender= df.Gender.fillna(df.Gender.mode().iloc[0]) #Assigning mode values to NAN gender
display(df.loc[~((df.Gender=='Male')|(df.Gender=='Female'))])
display(df[df.Gender.isna()==True])
```

Cleaning Emails

```python
display(df.Email.info())

print(df['Email'].isnull().sum())
```

```python
df['Email'] = df['Email'].str.strip()
df['Email'] = df['Email'].str.lower()
pattern = r'^[a-z0-9._%+-]+@[a-z0-9.-]+\.[a-z]{2,}$'
invalid_emails = df.loc[~df['Email'].str.match(pattern, na=False)]
display(invalid_emails)


```

```python
df.loc[~df['Email'].str.match(pattern, na=False),'Email'] = "Not Available" #Filled missing and invalid emails with Not Available
display(df.Email.info())

print(df['Email'].isnull().sum())
```

Cleaning Phone Numbers

```python
display(df.Phone.info())
df.Phone.isnull().sum()
```

```python
df.Phone=df.Phone.replace(r'\D', '', regex=True) #Remove non digit values in numbers
df['Phone'] = df['Phone'].str.replace(r'^91(?=\d{10}$)', '', regex=True) #trim country code

df['Phone'] = df['Phone'].replace(r'^\s*$', np.nan, regex=True)
df.loc[df['Phone'].isna() | (df['Phone'].str.len() < 10), 'Phone'] = 'Unknown'
df[df.Phone.str.len()!=10].head(5) #Replaced all NAN & inalid numbers to unknown

```

Convert date columns to date format

```python
#df['Purchase_Date']=pd.to_datetime(df.Purchase_Date,errors='coerce')
s = df["Purchase_Date"].astype("string").str.strip()
parsed = pd.Series(pd.NaT, index=df.index, dtype="datetime64[ns]")

formats = {
    r"^\d{2}/\d{2}/\d{4}$": "%d/%m/%Y",  # 31/05/2023
    r"^\d{2}-\d{2}-\d{4}$": "%m-%d-%Y",  # 06-26-2024
    r"^\d{4}-\d{2}-\d{2}$": "%Y-%m-%d",  # 2023-04-11
}

for pattern, date_format in formats.items():
    mask = s.str.match(pattern, na=False)
    parsed.loc[mask] = pd.to_datetime(
        s[mask], format=date_format, errors="coerce"
    )

# Review any nonblank values that did not parse before replacing the column
unparsed = s.notna() & s.ne("") & parsed.isna()
print(df.loc[unparsed, "Purchase_Date"])
df["Purchase_Date"] = parsed.dt.strftime("%d/%m/%Y")
df["Purchase_Date"] = parsed
df.Purchase_Date=pd.to_datetime(df.Purchase_Date)
fill_date = df["Purchase_Date"].mode().iloc[0]
df["Purchase_Date"] = df["Purchase_Date"].fillna(fill_date)
df.info()
```

Standerdize City Names

```python
print(df.City.isna().sum())
display(df[df.City.isna()|df.City.isnull()])
```

```python
df.City=df.City.fillna("Not available") #Filling non values with not available
df.City.info()
```

```python
df.City=df.City.str.title() #Standardize city
df.City.head()
```

Detect outliers

```python
df.Purchase_Amount.describe()
```

```python
df["Purchase_Amount"] = pd.to_numeric(df["Purchase_Amount"], errors="coerce")
print(df["Purchase_Amount"].sort_values().head(10))
print(df["Purchase_Amount"].sort_values(ascending=False).head(10))



```

```python
median = df["Purchase_Amount"].median()
df["Purchase_Amount"] = df["Purchase_Amount"].fillna(median)
df.loc[df["Purchase_Amount"] > 200000, "Purchase_Amount"] = median
df.Purchase_Amount.describe()
```
