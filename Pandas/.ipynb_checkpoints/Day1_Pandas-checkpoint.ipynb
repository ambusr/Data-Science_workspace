```python
import pandas as pd
import numpy as np
```

```python
series = pd.Series(['New York', 'Chicago', 'Toronto', 'Lisbon'])

# Creating the row axis labels
series.index = ['City 1', 'City 1', 'City 3', 'City 3'] 
print(series['City 1'])
```

```python
# Assign values to Data, Index Assign attributes
name=np.array(['Raju','Ram','Abhi','Siju'])
mark=np.array([20,50,55,100])
Ser=pd.Series(data=name,index=mark)
print(Ser)
```

```python
# Dictionary to pandas series
demo_dict={'Arun':30,'Kannan':55,'visakh':100}
pd.Series(demo_dict)
```

```python
#Creating a Series Using NumPy Functions
LinSpace_demo=pd.Series(np.linspace(1,10,5))
Range_Demo=pd.Series(range(1,10))
Arange_Demo=pd.Series(np.arange(1,10))
print(f"Using LineSpace fn:\n{LinSpace_demo},\nUsing Range Fn:\n{Range_Demo}\nUsing ARange Fn:\n{Arange_Demo}")
```

```python
#Accessing Elements 
ser=pd.Series(['P','A','N','D','A','S'],index=[1,2,3,4,5,6])

print(ser[1])
print(ser[6])
print(ser[[1,3,5]])
```

```python
#Slicing using above ser Series

print(ser[:6])
print(ser[0:])
print(ser[-6:])
```

```python
import pandas as pd
s1 = pd.Series([True, False, True])
s2 = pd.Series([False, False, True])

# Applying logical AND
result = s1 & s2
print(result)
```

```python
# Index & Data modification
PanSer=pd.Series(['Arun','Varun','Sanu','King','Saji','Anil','Sumesh'],index=[10,20,30,40,50,60,70])
print(PanSer.index)
#print(PanSer.data)
```
