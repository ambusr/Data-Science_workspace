```python
#handling missing data

#Topic 1 Check missing value
#isnull() returns a DataFrame of Boolean value where True represents missing data (NaN). 
#This is simple if we want to find and fill missing data in a dataset.

import pandas as pd
import numpy as np

d = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, 45, 56, np.nan],
        'Third Score': [np.nan, 40, 80, 98]}
df = pd.DataFrame(d)

mv = df.isnull()

print(mv)
```

```python
###Droppping missing values
## important


import pandas as pd
import numpy as np

dict = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, np.nan, 45, 56],
        'Third Score': [52, 40, 80, 98],
        'Fourth Score': [np.nan, np.nan, np.nan, 65]}
df = pd.DataFrame(dict)
print(df)
df.dropna()
```

```python
#dropping rows with all null values

dict = {'First Score': [100, np.nan, np.nan, 95],
        'Second Score': [30, np.nan, 45, 56],
        'Third Score': [52, np.nan, 80, 98],
        'Fourth Score': [np.nan, np.nan, np.nan, 65]}
df = pd.DataFrame(dict)
print(df)

df.dropna(how='all')
```

```python
#Drop colums that contain atleast one value

dict = {'First Score': [100, np.nan, np.nan, 95],
        'Second Score': [30, np.nan, np.nan, 56],
        'Third Score': [52, np.nan, 80, 98],
        'Fourth Score': [60, 67, 68, 65]}
df = pd.DataFrame(dict)
print(df)


print(df.dropna(axis=0))   #rows
print(df.dropna(axis=1))   #colums
```
