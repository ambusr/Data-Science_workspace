```python
import pandas as pd

data = {
    'EmpID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Amit', 'Priya', 'Rahul', 'Sneha', 'John',
             'Anu', 'Kiran', 'David', 'Ravi', 'Meena'],
    'Age': [25, 30, 28, 35, 40, 24, 29, 32, 27, 31],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR',
                   'IT', 'Finance', 'IT', 'HR', 'Finance'],
    'Salary': [45000, 55000, 60000, 70000, 80000,
               40000, 65000, 75000, 50000, 72000],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Chennai', 'Mumbai',
             'Delhi', 'Chennai', 'Mumbai', 'Delhi', 'Chennai']
        }
df = pd.DataFrame(data)
```

```python
# 1.Display the first 5 records.
print(df.iloc[0:5])
```

```python
# 2.Display the last 3 records.
print(df.iloc[-3:])
```

```python
# 3.Find the total number of rows and columns.
print(df.shape)
# or
rows,columns=df.shape
print(f"Total Rows:{rows} & Total Columns:{columns}")
```

```python
# 4.Display only Name, Salary and Department columns.
display(df[['Name','Salary','Department']])
```

```python
# 5.Find employees whose salary is greater than 50,000.
Emp_Sal=df[df.Salary>50000]
display(Emp_Sal)
```

```python
# 6.Find employees whose age is between 25 and 30.
Emp_age=df[(df.Age>=25) & (df.Age<=30)]
display(Emp_age)
```

```python
# 7.Display employees who belong to the IT department.
Emp_dep=df[df.Department=='IT']
print(Emp_dep)
```

```python
# 8.Sort employees by salary in descending order.
Emp_sort=df.sort_values('Salary',ascending=False)
display(Emp_sort)
```

```python
# 9.Find the average salary of all employees.
avg_Sal=df.Salary.sum()/df.Salary.shape[0]
print("Using calculation,Average:",avg_Sal)
print("Using Mean function. Average: ",df.Salary.mean())
```

```python
# 10.Find the maximum and minimum salary.
print(f"Maximum Salary: {df.Salary.max()} and Minimum Salaray:{df.Salary.min()}")
```

```python
# 11.Count employees department-wise.
print(df.groupby('Department').size())
```

```python
# 12.Calculate average salary for each department.
print(df.groupby('Department')['Salary'].mean())
```

```python
# 14.Find the second-highest salary.
print(df[df.Salary==df.Salary.max()])
```
