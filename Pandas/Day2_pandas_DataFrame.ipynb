```python
import pandas as pd
import warnings

warnings.filterwarnings(
    'ignore',
    message='The inplace keyword in DataFrame.set_index is deprecated'
)
```

```python
#Set index
data = {'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'], 
        'Age': [25, 30, 22, 35, 28], 
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'], 
        'Salary': [50000, 55000, 40000, 70000, 48000]}

df = pd.DataFrame(data)

df_with_index = df.set_index('Name')
print(df_with_index)
```

```python
player= [['M.S.Dhoni', 36, 75, 5428000],['A.B.D Villers', 38, 74, 3428000],['V.Kohli', 31, 70, 8428000],['S.Smith', 34, 80, 4428000],['C.Gayle', 40, 100, 4528000],
['J.Root', 33, 72, 7028000],['K.Peterson', 42, 85, 2528000]]
player_list=pd.DataFrame(player,columns=['Name','age','TopScore','Salary'])
#Accessing Data
print(player_list)
print("Accessing specific data using iloc",player_list.iloc[2,3])
player_list.set_index('Name',inplace=True)
print("Acessing specific data using loc",player_list.loc['C.Gayle','Salary'])
```
